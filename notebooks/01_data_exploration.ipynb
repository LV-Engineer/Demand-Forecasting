# Cell data

```python
import pandas as pd
from demand.config import RAW

train = pd.read_csv(RAW / "train.csv", parse_dates=["date"])
stores = pd.read_csv(RAW / "stores.csv")
print(train.date.min(), train.date.max(), train.date.nunique())
train.head()
```

# Product families (last ~2 years)

```python
recent = train[train.date >= "2015-08-16"]
fam = recent.groupby("family").sales.agg(
    total="sum", zero_share=lambda s: (s == 0).mean()
)
fam.sort_values("total", ascending=False).round(2)
```

# Stores overview

```python
first_date = train.groupby("store_nbr").date.min()
tot = recent.groupby("store_nbr").sales.sum().rename("total")
info = stores.set_index("store_nbr").join(tot).join(first_date.rename("first_date"))
info.sort_values("total", ascending=False).head(15)
```
